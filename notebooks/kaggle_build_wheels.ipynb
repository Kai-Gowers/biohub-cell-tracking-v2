# Kaggle: build an offline wheelhouse for the submission notebook

Run this ONCE, with Settings > Internet ON, then click "Save Version" -- the `/kaggle/working/wheels` folder becomes this notebook's Output, which Kaggle lets you attach to other notebooks as a Dataset (e.g. name it `cell-tracking-wheels`).

Why this exists: `kaggle_run.ipynb` is the actual submission notebook and runs with internet disabled during the official scored rerun, regardless of this notebook's own settings. `blosc2`/`zarr`/`zstandard` are not part of Kaggle's default image, so without a pre-built wheelhouse attached, the submission notebook's import of `cell_tracking.config` (which pulls in all three transitively) fails with no way to recover.

Downloading here rather than on a dev machine matters: `pip download` resolves wheels for the CURRENT interpreter and platform, so running it in a Kaggle notebook guarantees the wheels match the actual submission environment -- a local Mac/Linux dev machine would fetch the wrong platform tag.

```python
import subprocess, sys
from pathlib import Path

WHEELS_DIR = Path('/kaggle/working/wheels')
WHEELS_DIR.mkdir(parents=True, exist_ok=True)

subprocess.run(
    [sys.executable, '-m', 'pip', 'download', '--dest', str(WHEELS_DIR),
     'blosc2', 'zarr', 'zstandard'],
    check=True,
)

wheels = sorted(p.name for p in WHEELS_DIR.iterdir())
print(f'{len(wheels)} files in {WHEELS_DIR}:')
for w in wheels:
    print(' ', w)
```

Now: **Save Version** (this saves `/kaggle/working/wheels` as the notebook's Output), then in `kaggle_run.ipynb` attach this notebook's output as a Dataset alongside the code and competition data.
